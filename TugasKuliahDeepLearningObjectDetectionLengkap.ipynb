```python
# Tugas 1 - Download Pre-Trained YOLO
from ultralytics import YOLO
from PIL import Image

# 1. Load a model (Menggunakan YOLOv11s sesuai contoh slide 26), dipilih YOLO yang S (Small)
model = YOLO("yolo11s.pt")

# 2. Perform object detection on an image
# Pastikan nama file ini sama dengan gambar yang Anda upload di Langkah 3
results = model("Gambar_Demo.jpg")

# 3. Menampilkan hasil di Google Colab (Pengganti fungsi .show())
im_array = results[0].plot()
im = Image.fromarray(im_array[..., ::-1])
display(im)
```

```python
#Coba mengambil gambar lebih banyak, kita perkecil confident levelnya jadi 0.15

from ultralytics import YOLO
from PIL import Image

# Memuat model yang sama
model = YOLO("yolo11s.pt")

# Menjalankan deteksi dengan penyesuaian parameter
results = model(
    "Gambar_Demo.jpg",
    conf=0.15,    # Menurunkan batas keyakinan (Confidence) menjadi 15%
    imgsz=1024,   # Memperbesar ukuran gambar yang diproses agar detail orang di belakang terlihat
    iou=0.6       # Mengizinkan kotak deteksi lebih banyak bertumpuk (Intersection over Union)
)

# Menampilkan hasil di Google Colab
im_array = results[0].plot()
im = Image.fromarray(im_array[..., ::-1])
display(im)
```

```python
#Kita coba perbesar image resolutionnya jadi 1280, kemudian confident level kita tingkatkan di 0.7

from ultralytics import YOLO
from PIL import Image

# 1. Mengganti model dari Small (yolo11s.pt) menjadi Medium (yolo11m.pt)
# Model akan otomatis diunduh oleh Colab (ukurannya lebih besar, tapi lebih akurat)
model = YOLO("yolo11m.pt")

# 2. Deteksi dengan syarat ketat (Minimal 60% yakin)
results = model(
    "Gambar_Demo.jpg",
    conf=0.7,       # Syarat keyakinan minimal 60%
    imgsz=1280,     # Memperbesar resolusi gambar menjadi sangat tinggi (1280)
    iou=0.6,        # Mengizinkan kotak yang berdempetan di kerumunan
    classes=[0]     # Angka 0 adalah kode khusus untuk hanya mendeteksi "Person" (Orang)
)

# 3. Menampilkan hasil di Google Colab
im_array = results[0].plot()
im = Image.fromarray(im_array[..., ::-1])
display(im)
```
