```python
import glob
from ultralytics import YOLO

# 1. Mencari lokasi asli folder secara pasti (mengatasi masalah folder ganda)
train_dir = glob.glob('/content/**/train/images', recursive=True)[0]
val_dir = (glob.glob('/content/**/valid/images', recursive=True) + glob.glob('/content/**/val/images', recursive=True))[0]

# 2. Menulis file konfigurasi baru dengan alamat yang sudah pasti benar
yaml_text = f"train: {train_dir}\nval: {val_dir}\nnc: 3\nnames: ['head', 'helmet', 'person']"

with open("/content/data_tugas_final.yaml", "w") as f:
    f.write(yaml_text)

print(f"✅ Gambar Train berhasil ditemukan di: {train_dir}")
print(f"✅ Gambar Valid berhasil ditemukan di: {val_dir}")
print("🚀 Memulai proses training model...\n")

# 3. Menjalankan proses training
model = YOLO("yolo11n.pt")
model.train(
    data="/content/data_tugas_final.yaml",
    epochs=25,
    imgsz=640,
    device="cpu"
)
print("\n✅ Langkah (2) selesai! Model berhasil dilatih.")
```
