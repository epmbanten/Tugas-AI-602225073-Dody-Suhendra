```python
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.applications.efficientnet import preprocess_input, decode_predictions
import matplotlib.pyplot as plt # Digunakan untuk menampilkan gambar di Colab

# 1. Load Model [cite: 803, 807]
model = EfficientNetB0(weights='imagenet')

# 2. Baca Gambar
frame = cv2.imread('test.jpg')

# 3. Preprocessing
image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
image_resized = cv2.resize(image_rgb, (224, 224))
image_array = np.expand_dims(image_resized, axis=0)
image_array = preprocess_input(image_array)

# 4. Klasifikasi/Inference
predictions = model.predict(image_array)
_, label, prob = decode_predictions(predictions, top=1)[0][0]

# 5. Tampilkan Hasil
print(f"Hasil Klasifikasi: {label} ({prob:.2f})")
plt.imshow(image_rgb)
plt.title(f"Label: {label}, Prob: {prob:.2f}")
plt.axis('off')
plt.show()
```
